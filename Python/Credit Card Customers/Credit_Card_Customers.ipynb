```python
fig, axs = plt.subplots(2, 2, figsize=(16,12))

axs[0,0].hist(bank["Customer_Age"], bins=20, color="skyblue", edgecolor='black')
axs[0,0].set_title("Customer Age Distribution")
axs[0,0].set_xlabel("Age")
axs[0,0].set_ylabel("Frequency")

edu_counts = bank["Education_Level"].value_counts()
bars_edu = axs[0,1].bar(edu_counts.index, edu_counts.values, color="lightgreen", edgecolor='black')
axs[0,1].set_title("Distribution of Education Level")
axs[0,1].set_xlabel("Education Level")
axs[0,1].set_ylabel("Number of Customers")
axs[0,1].tick_params(axis='x', rotation=30)
axs[0,1].bar_label(bars_edu, padding=3)

income_counts = bank["Income_Category"].value_counts()
bars_income = axs[1,0].bar(income_counts.index, income_counts.values, color="lightblue", edgecolor='black')
axs[1,0].set_title("Distribution of Customers by Income Category")
axs[1,0].set_xlabel("Income Category")
axs[1,0].set_ylabel("Number of Customers")
axs[1,0].tick_params(axis='x', rotation=30)
axs[1,0].bar_label(bars_income, padding=3)

card_counts = bank["Card_Category"].value_counts()
bars_card = axs[1,1].bar(card_counts.index, card_counts.values, color="salmon", edgecolor='black')
axs[1,1].set_title("Number of Customers by Card Category")
axs[1,1].set_xlabel("Card Category")
axs[1,1].set_ylabel("Number of Customers")
axs[1,1].bar_label(bars_card, padding=3)

plt.tight_layout()
plt.show()
```

```python

counts = bank.groupby(['Card_Category','Income_Category']).size().unstack(fill_value=0)
counts.plot(kind='bar', figsize=(10,5), width=0.8, color=['#66b3ff','#ff9999','#99ff99','#ffcc99','#636CCB','#A376A2'])
plt.title("Card Category vs Income Category"); plt.xlabel("Card Category"); plt.ylabel("Number of Customers")
plt.tight_layout();
plt.show()
```

```python

fig, axs = plt.subplots(2, 2, figsize=(15,10))

card_counts = bank['Card_Category'].value_counts()
bars1 = axs[0,0].bar(card_counts.index, card_counts.values, color='lightgreen', edgecolor='black')
axs[0,0].set_title('Number of Customers by Card Category')
axs[0,0].set_ylabel('Count')
axs[0,0].bar_label(bars1, padding=3)

card_credit_avg = bank.groupby('Card_Category')['Credit_Limit'].mean()
bars2 = axs[0,1].bar(card_credit_avg.index, card_credit_avg.values, color='lightblue', edgecolor='black')
axs[0,1].set_title('Average Credit Limit by Card Category')
axs[0,1].set_ylabel('Average Credit Limit')
axs[0,1].bar_label(bars2, padding=3, fmt='%.0f')

card_util_avg = bank.groupby('Card_Category')['Avg_Utilization_Ratio'].mean()
bars3 = axs[1,0].bar(card_util_avg.index, card_util_avg.values, color='salmon', edgecolor='black')
axs[1,0].set_title('Average Utilization Ratio by Card Category')
axs[1,0].set_ylabel('Avg Utilization Ratio')
axs[1,0].bar_label(bars3, padding=3, fmt='%.2f')

card_trans_avg = bank.groupby('Card_Category')['Total_Trans_Amt'].mean()
bars4 = axs[1,1].bar(card_trans_avg.index, card_trans_avg.values, color='plum', edgecolor='black')
axs[1,1].set_title('Average Total Transactions by Card Category')
axs[1,1].set_ylabel('Average Transactions Amount')
axs[1,1].bar_label(bars4, padding=3, fmt='%.0f')

plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(14,6))

labels = bank["Attrition_Flag"].value_counts().index
sizes = bank["Attrition_Flag"].value_counts().values
axs[0].pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=["#66b3ff", "#ff6666"])
axs[0].set_title("Customer Attrition Rate")

counts = bank.groupby(['Gender','Attrition_Flag']).size().unstack()
counts.plot(kind='bar', color=['#ff6666','#66b3ff'], ax=axs[1])
axs[1].set_title("Customer Attrition by Gender")
axs[1].set_xlabel("Gender")
axs[1].set_ylabel("Count")
axs[1].legend(title="Attrition_Flag")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x="Attrition_Flag", y="Total_Trans_Ct", data=bank, palette="Set3")
plt.title("Boxplot of Total Transactions Count by Customer Status")
plt.xlabel("Customer Status")
plt.ylabel("Total Transactions Count")
plt.tight_layout()
plt.show()
```

```python
means = bank.groupby(['Attrition_Flag','Card_Category'])['Contacts_Count_12_mon'].mean().unstack()
means.plot(kind='bar', figsize=(10,6))
plt.xlabel("Customer Status")
plt.ylabel("Average Contacts")
plt.title("Contacts by Status & Card Category")
plt.legend(title="Card Category")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

selected_cols = ['Customer_Age', 'Credit_Limit', 'Total_Trans_Amt', 'Total_Trans_Ct', 'Avg_Utilization_Ratio']

corr = bank[selected_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", ax=ax, annot_kws={"size":10})

ax.set_title("Correlation Heatmap")
plt.tight_layout()
plt.show()

```
